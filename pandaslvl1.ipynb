Find the top 10 bowlers with the lowest economy rate, subject to these conditions:
1. Consider only bowlers who have bowled at least 500 legal deliveries.
2. Calculate total runs conceded by each bowler.
3. Calculate total legal deliveries bowled.
4. Calculate economy rate:
\[
\text{Economy} = \frac{\text{Runs conceded}}{\text{Legal balls}} \times 6
\]

5. Sort by economy rate in ascending order.
6. Display the top 10 bowlers.

```python
without_extra=ipld[(ipld['noball_runs'] == 0) & (ipld['wide_runs'] == 0)]
ball_count=without_extra.groupby("bowler")['ball'].count().sort_values(ascending=False)
legal_balls=ball_count[ball_count >= 500]
runs=ipld.groupby('bowler')['total_runs'].sum().sort_values(ascending=False)
runs=runs.loc[legal_balls.index]
economy_rate=(runs/legal_balls)*6
economy_rate.sort_values().head(10)
```

```python
runs=runs.rename("Runs Concedered")
legal_balls=legal_balls.rename("Legal Balls")
economy_rate=economy_rate.rename("Economy Rate")
```

```python
result=pd.concat([runs,legal_balls,economy_rate],axis=1)
result=result.reset_index()
result = result.rename(columns={'index': 'Bowler'})
result
```

**IPL Matches – Team Performance Dashboard
Dataset: ipl (matches dataset)

Objective: Build a team-wise performance table for all teams.

Calculate these 5 metrics:
1. Matches Played: Total matches in which the team appeared (Team1 or Team2).
2. Matches Won: Total matches won (WinningTeam).
3. Win Percentage: Matches Won / Matches Played × 100.
4. Toss Wins: Total tosses won (TossWinner).
5. Toss-to-Match Win Conversion: Matches won after winning the toss / Total tosses won × 100.**

```python
matches_played=ipl['Team1'].value_counts()+ipl['Team2'].value_counts()
matches_won=ipl['WinningTeam'].value_counts()
win=(matches_won/matches_played)*100
win=win.sort_values(ascending=False)
toss_won=ipl['TossWinner'].value_counts()
toss_to_match_won=ipl[ipl['TossWinner']==ipl['WinningTeam']].value_counts('TossWinner')
toss_conversion=(toss_to_match_won/toss_won)*100
toss_conversion=toss_conversion.sort_values(ascending=False)
matches_played=matches_played.rename("Matches Played")
matches_won=matches_won.rename("Matches Won")
win=win.rename("Win%")
toss_won=toss_won.rename('Toss Won')
toss_to_match_won=toss_to_match_won.rename("Toss To Match Won")
toss_conversion=toss_conversion.rename("Toss Conversion")
result=pd.concat([matches_played,matches_won,win,toss_won,toss_to_match_won,toss_conversion],axis=1)
result=result.fillna(0)
result.sort_values('Win%',ascending=False).head(10)
```

**Movie Genre Performance Analysis 🎯
Intermediate → Advanced
Dataset: movies

Build a genre-wise analysis table with these metrics:
1. Movie Count
2. Average Rating
3. Median Rating
4. Highest Rating
5. Lowest Rating
Conditions:
- Exclude rows with missing genres or imdb_rating.
- A movie belonging to multiple genres must count in every relevant genre.
- Include only genres with at least 30 movies.
- Sort by Average Rating in descending order.
- Display the top 10 genres.
Expected output: One row per genre, with all five metrics.**

```python
drop_movies=movies[['genres','imdb_rating']].dropna()
drop_movies['genres']=drop_movies['genres'].str.split('|')
drop_movies=drop_movies.explode('genres')
completed_movies=drop_movies.groupby("genres").agg(
    Movies_count=('imdb_rating','count'),
    Average_rating=('imdb_rating','mean'),
    Median_rating=('imdb_rating','median'),
    Highest_rating=('imdb_rating','max'),
    Lowest_rating=('imdb_rating','min')
)
result=completed_movies[completed_movies['Movies_count'] >= 30].sort_values(by='Average_rating',ascending=False)
result.head(10)
```

Challenge 12: IPL Batsman Performance Analysis
Advanced
Dataset: ipld (IPL deliveries)

Build a batsman performance table with these metrics:
1. Total Runs
2. Total Balls Faced
3. Number of Fours
4. Number of Sixes
5. Strike Rate
6. Total Innings Played
Conditions:
- Exclude wides when counting balls faced.
- Count fours using batsman_runs == 4.
- Count sixes using batsman_runs == 6.
- Strike Rate = (Total Runs / Total Balls Faced) × 100.
- Include only batsmen with at least 500 balls faced.
- Sort by Strike Rate descending.
- Display the top 10.

```python
batsman_total_runs=ipld.groupby('batsman')['batsman_runs'].sum()
balls_count=ipld[ipld['wide_runs']==0]
balls_count=balls_count.groupby('batsman')['ball'].count()
fours=ipld[(ipld['batsman_runs']==4) ]
sixes = ipld[ipld['batsman_runs'] == 6]
no_of_fours=fours.groupby('batsman').size()
no_of_sixes=sixes.groupby('batsman').size()
strike_rate=(batsman_total_runs/balls_count)*100
total_innings_played=ipld[['match_id','inning','batsman']].drop_duplicates()
total_innings_played=total_innings_played.groupby('batsman')['inning'].size()
batsman_total_runs=batsman_total_runs.rename("Total Runs")
balls_count=balls_count.rename("Total Balls Faced")
no_of_fours=no_of_fours.rename('No of Fours')
no_of_sixes=no_of_sixes.rename("No of sixes")
strike_rate=strike_rate.rename("StrikeRate")
total_innings_played=total_innings_played.rename("Total Innings Played")
result=pd.concat([batsman_total_runs,balls_count,no_of_fours,no_of_sixes,strike_rate,total_innings_played],axis=1)
result = result[result['Total Balls Faced'] >= 500]
result.sort_values(by='StrikeRate',ascending=False).head(10)
```

IPL Bowling Performance Analysis
Dataset: ipld

Build a bowler-wise performance table with these six metrics:
1. Total Wickets
2. Runs Conceded
3. Legal Balls Bowled
4. Bowling Average = Runs Conceded / Total Wickets
5. Economy Rate = Runs Conceded / Legal Balls × 6
6. Bowling Strike Rate = Legal Balls / Total Wickets
Conditions:
- Exclude run out, retired hurt, obstructing the field, and retired out from bowler wickets.
- Exclude wides and no-balls when counting legal balls.
- Count runs conceded from all deliveries using total_runs.
- Include only bowlers with at least 500 legal balls.
- Sort by Bowling Average ascending.
- Display the top 10 bowlers.

```python
ipld.columns
```

```python
filtered_wickets=ipld[((ipld['dismissal_kind'] !='run out') & (ipld['dismissal_kind'] != 'retired hurt')) & ( (ipld['dismissal_kind'] != 'obstructing the field') & (ipld['dismissal_kind']!='retired out')) ]
total_wickets=filtered_wickets.groupby('bowler')['dismissal_kind'].count()
runs_conceded=ipld.groupby('bowler')['total_runs'].sum()
legal_balls=ipld[(ipld['wide_runs']==0) & (ipld['noball_runs'] == 0)]
legal_balls=legal_balls.groupby('bowler')['ball'].count()
bowling_average=runs_conceded/total_wickets
economy_rate=(runs_conceded/legal_balls)*6
bowler_strike_rate=legal_balls/total_wickets
total_wickets=total_wickets.rename("Total Wickets")
runs_conceded=runs_conceded.rename("Runs Conceded")
legal_balls=legal_balls.rename("Legal Balls")
bowling_average=bowling_average.rename("Bowling Average")
economy_rate=economy_rate.rename("Economy Rate")
bowler_strike_rate=bowler_strike_rate.rename("Bowler Strike Rate")
result=pd.concat([total_wickets,runs_conceded,legal_balls,bowling_average,economy_rate,bowler_strike_rate],axis=1)
result=result[result['Legal Balls'] >= 500]
result.sort_values(by='Bowling Average', ascending=True).head(10)
```
